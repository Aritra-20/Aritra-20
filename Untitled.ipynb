```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('/content/sample_data/credit_Assn1_SMDM.csv')
display(df.head())
```

###Visual Summaries
Histograms for numerical variables and bar charts for categorical variables to understand the distributions.

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
sns.histplot(df['Income'], kde=True, ax=axes[0, 0]).set_title('Income Distribution')
sns.histplot(df['Limit'], kde=True, ax=axes[0, 1]).set_title('Limit Distribution')
sns.countplot(data=df, x='Region', ax=axes[1, 0]).set_title('Region Count')
sns.countplot(data=df, x='Own', ax=axes[1, 1]).set_title('Home Ownership Count')
plt.tight_layout()
plt.show()
```

Q1. Make suitable visual summaries for the dataset. Comment.

---


Ans: Histograms show that Income and Limit are right-skewed, indicating most customers are in the lower-to-middle range with a few high-value exceptions. The South region has the highest concentration of customers.

### Income Statistics

```python
income_stats = df['Income'].describe()
iqr_income = income_stats['75%'] - income_stats['25%']

print(f"Mean Income: {income_stats['mean']:.2f}")
print(f"Q1: {income_stats['25%']:.2f}, Q2 (Median): {income_stats['50%']:.2f}, Q3: {income_stats['75%']:.2f}")
print(f"Standard Deviation: {income_stats['std']:.2f}")
print(f"IQR: {iqr_income:.2f}")
```

### Contingency Table (Card Ownership vs Region)

```python
df['Card_Category'] = df['Cards'].apply(lambda x: 'low' if x <= 3 else 'high')
contingency_table = pd.crosstab(df['Card_Category'], df['Region'])
display(contingency_table)
```
